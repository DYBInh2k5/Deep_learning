## Bài 5. Tính gradient bằng `tf.GradientTape()`

Yêu cầu:

1. Tạo biến `w = 3.0` bằng `tf.Variable`.
2. Xét hàm `y = w^2 + 2w + 1`.
3. Dùng `tf.GradientTape()` để tính gradient `dy/dw`.
4. In ra giá trị của `y` và gradient thu được.

```python
# TODO: Tạo biến w
w = tf.Variable(3.0)

# TODO: Dùng GradientTape để tính y và gradient dy/dw
with tf.GradientTape() as tape:
    y = w**2 + 2*w + 1
# TODO: In ra giá trị y
dy_dw = tape.gradient(y, w)

# TODO: In ra gradient dy/dw
print("Giá trị của y:", y.numpy())
print("Gradient dy/dw:", dy_dw.numpy())
```

## Bài 6. Mô phỏng gradient descent đơn giản

Yêu cầu:

Xét hàm mất mát: $loss = (w - 4.0)^2$

Thực hiện gradient descent trong 10 bước để cập nhật `w` sao cho `w` tiến dần về 4.

Yêu cầu cụ thể:

1. Khởi tạo `w = 0.0`
2. Đặt `learning_rate = 0.1`
3. Lặp 10 bước
4. Mỗi bước in ra: số bước, giá trị `w`, giá trị `loss`, gradient

```python
# TODO: Khởi tạo w và learning_rate
w = tf.Variable(0.0)
learning_rate = 0.1

# TODO: Viết vòng lặp gradient descent 10 bước
for step in range(10):
    with tf.GradientTape() as tape:
        loss = (w - 4.0) ** 2
    grad = tape.gradient(loss, w)
    w.assign_sub(learning_rate * grad)
    print(f"Step {step+1}: w = {w.numpy():.4f}, loss = {loss.numpy():.4f}, grad = {grad.numpy():.4f}")
```

## Bài 7. Xây dựng mô hình Keras đơn giản

Yêu cầu:

Tạo mô hình `Sequential` gồm:

1. Một layer `Dense` có 16 neuron, activation=`relu`, input shape = `(8,)`
2. Một layer output `Dense` có 1 neuron, activation=`sigmoid`
3. Hiển thị `model.summary()`

```python
# TODO: Tạo mô hình Sequential theo yêu cầu
model = keras.Sequential([
    layers.Dense(16, activation='relu', input_shape=(8,)),
    layers.Dense(1, activation='sigmoid')
])
# TODO: Hiển thị summary của mô hình
model.summary()
```

## Bài 8. Tạo dữ liệu giả cho bài toán phân loại nhị phân

Yêu cầu:

1. Dùng NumPy tạo `x_train` gồm 200 mẫu, mỗi mẫu có 8 đặc trưng ngẫu nhiên.
2. Tạo `y_train` theo quy tắc: nếu tổng 8 đặc trưng > 4 thì gán nhãn 1, ngược lại nhãn 0.
3. Tạo tương tự `x_test` gồm 50 mẫu và `y_test`.
4. In shape của `x_train`, `y_train`, `x_test`, `y_test`.

```python
# TODO: Đặt random seed để kết quả có thể tái lập


# TODO: Tạo x_train và y_train
x_train = np.random.rand(200, 8)
y_train = (np.sum(x_train, axis=1) > 4).astype(int)

# TODO: Tạo x_test và y_test
x_test = np.random.rand(50, 8)
y_test = (np.sum(x_test, axis=1) > 4).astype(int)

# TODO: In shape của các tập dữ liệu
print("x_train shape:", x_train.shape)
print("y_train shape:", y_train.shape)
print("x_test shape:", x_test.shape)
print("y_test shape:", y_test.shape)
```

## Bài 9. Compile và huấn luyện mô hình

Yêu cầu:

1. Compile model với:
   - optimizer = `rmsprop`
   - loss = `binary_crossentropy`
   - metrics = `accuracy`
2. Train mô hình với:
   - epochs = 5
   - batch_size = 16
   - validation_split = 0.2
3. Lưu lịch sử train vào biến `history`.

```python
# TODO: Compile mô hình
model.compile(
    optimizer='rmsprop',
    loss='binary_crossentropy',
    metrics=['accuracy'])



# TODO: Huấn luyện mô hình và lưu vào biến history
history = model.fit(
    x_train, y_train,
    epochs=5,
    batch_size=16,
    validation_split=0.2)



```

## Bài 10. Đánh giá mô hình và dự đoán

Yêu cầu:

1. Đánh giá mô hình trên tập test.
2. In ra `test_loss` và `test_accuracy`.
3. Dự đoán xác suất cho 5 mẫu đầu tiên của `x_test`.
4. In kết quả dự đoán.
5. Hãy cho biết: nếu output sigmoid >= 0.5 thì thuộc lớp nào?

```python
# TODO: Đánh giá mô hình trên tập test
test_loss, test_accuracy = model.evaluate(x_test, y_test)

# TODO: In test_loss và test_accuracy
print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)

# TODO: Dự đoán xác suất cho 5 mẫu đầu tiên
predictions = model.predict(x_test[:5])

# TODO: In kết quả dự đoán
print("Dự đoán (xác suất):")
print(predictions)
```
